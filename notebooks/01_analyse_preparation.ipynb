```python
import pandas as pd

fichier = "../data/raw/Online Retail.xlsx"

df = pd.read_excel(fichier)

print("Dataset chargé :", df.shape)
```

```python
# 1. Transactions avec une quantité négative
quantites_negatives = (df["Quantity"] < 0).sum()

# 2. Transactions avec une quantité nulle
quantites_nulles = (df["Quantity"] == 0).sum()

# 3. Transactions avec un prix unitaire nul
prix_nuls = (df["UnitPrice"] == 0).sum()

# 4. Transactions avec un prix unitaire négatif
prix_negatifs = (df["UnitPrice"] < 0).sum()

# 5. Factures d'annulation (InvoiceNo commençant par C)
annulations = df["InvoiceNo"].astype(str).str.startswith("C")

nombre_lignes_annulation = annulations.sum()
nombre_factures_annulation = df.loc[annulations, "InvoiceNo"].nunique()

# Affichage
print("ANOMALIES COMMERCIALES")
print("Quantités négatives :", quantites_negatives)
print("Quantités nulles :", quantites_nulles)
print("Prix unitaires nuls :", prix_nuls)
print("Prix unitaires négatifs :", prix_negatifs)
print("Lignes d'annulation :", nombre_lignes_annulation)
print("Factures d'annulation distinctes :", nombre_factures_annulation)
```

```python
# Identifier les factures d'annulation
mask_annulation = df["InvoiceNo"].astype(str).str.startswith("C")

# Identifier les prix négatifs
mask_prix_negatif = df["UnitPrice"] < 0

# Identifier les prix nuls
mask_prix_nul = df["UnitPrice"] == 0

# Examiner les deux lignes avec un prix négatif
print("1. TRANSACTIONS AVEC UN PRIX NÉGATIF")
display(
    df.loc[
        mask_prix_negatif,
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID"]
    ]
)

# Examiner quelques lignes avec un prix nul
print("\n2. EXEMPLES DE TRANSACTIONS AVEC UN PRIX NUL")
display(
    df.loc[
        mask_prix_nul,
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice", "CustomerID"]
    ].head(10)
)

# Étudier le chevauchement entre quantités négatives et annulations
mask_quantite_negative = df["Quantity"] < 0

print("\n3. CHEVAUCHEMENT DES ANOMALIES")
print(
    "Quantités négatives associées à une annulation :",
    (mask_quantite_negative & mask_annulation).sum()
)
print(
    "Quantités négatives hors annulation :",
    (mask_quantite_negative & ~mask_annulation).sum()
)
```

```python
# Identifier les quantités négatives hors factures d'annulation
mask_retours = (df["Quantity"] < 0) & (~mask_annulation)

# Créer un DataFrame temporaire pour l'analyse
df_retours = df.loc[mask_retours].copy()

print("ANALYSE DES 1 336 LIGNES")

# 1. Vérifier les prix unitaires
print("\nRépartition des prix unitaires :")
print(df_retours["UnitPrice"].value_counts(dropna=False))

# 2. Vérifier les identifiants clients
print("\nIdentifiants clients manquants :")
print(df_retours["CustomerID"].isna().sum())

# 3. Examiner les descriptions
print("\nDescriptions les plus fréquentes :")
print(df_retours["Description"].value_counts(dropna=False).head(10))

# 4. Afficher quelques exemples
print("\nExemples de transactions :")
display(df_retours.head(15))
```

```python
# Identifier les lignes avec un prix nul
mask_prix_nul = df["UnitPrice"] == 0

# Exclure les 1 336 quantités négatives hors annulation
mask_autres_prix_nuls = mask_prix_nul & ~mask_retours

df_prix_nuls_restants = df.loc[mask_autres_prix_nuls].copy()

print("ANALYSE DES AUTRES PRIX NULS")
print("Nombre de lignes :", len(df_prix_nuls_restants))

print("\nRépartition des quantités :")
print(df_prix_nuls_restants["Quantity"].value_counts().head(15))

print("\nDescriptions les plus fréquentes :")
print(df_prix_nuls_restants["Description"].value_counts(dropna=False).head(15))

print("\nIdentifiants clients manquants :")
print(df_prix_nuls_restants["CustomerID"].isna().sum())

print("\nExemples de transactions :")
display(df_prix_nuls_restants.head(15))
```

```python
print("RÉPARTITION DES FACTURES")

# Les factures commencent-elles par C ou A ?
print("\nFactures commençant par C :")
print(
    df_prix_nuls_restants["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

print("\nFactures commençant par A :")
print(
    df_prix_nuls_restants["InvoiceNo"]
    .astype(str)
    .str.startswith("A")
    .sum()
)

print("\nRépartition des quantités :")
print(
    df_prix_nuls_restants["Quantity"]
    .apply(lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero")
    .value_counts()
)

print("\nDescriptions les plus fréquentes :")
print(
    df_prix_nuls_restants["Description"]
    .value_counts(dropna=False)
    .head(20)
)

print("\nNombre de factures distinctes :")
print(df_prix_nuls_restants["InvoiceNo"].nunique())
```

```python
# Regroupement des lignes par numéro de facture
analyse_factures = (
    df_prix_nuls_restants
    .groupby("InvoiceNo")
    .agg(
        nombre_lignes=("InvoiceNo", "size"),
        quantite_totale=("Quantity", "sum"),
        nombre_produits=("StockCode", "nunique"),
        nombre_clients=("CustomerID", "nunique"),
        date_facture=("InvoiceDate", "first"),
        pays=("Country", "first")
    )
    .sort_values("nombre_lignes", ascending=False)
)

print("Nombre de factures analysées :", len(analyse_factures))

print("\nLes 15 factures contenant le plus de lignes à prix nul :")
display(analyse_factures.head(15))
```

```python
# Sélectionner trois factures représentatives
factures_a_examiner = ["539856", "537534", "564651"]

# Récupérer toutes leurs lignes dans le dataset original
df_factures_completes = df[
    df["InvoiceNo"].astype(str).isin(factures_a_examiner)
].copy()

# Ajouter le montant calculé de chaque ligne
df_factures_completes["LineAmount"] = (
    df_factures_completes["Quantity"]
    * df_factures_completes["UnitPrice"]
)

# Résumé des factures
resume_factures = (
    df_factures_completes
    .groupby("InvoiceNo")
    .agg(
        nombre_total_lignes=("InvoiceNo", "size"),
        lignes_prix_nul=("UnitPrice", lambda x: (x == 0).sum()),
        lignes_prix_positif=("UnitPrice", lambda x: (x > 0).sum()),
        chiffre_affaires=("LineAmount", "sum"),
        client=("CustomerID", "first"),
        pays=("Country", "first")
    )
)

display(resume_factures)

# Afficher le détail des transactions
display(
    df_factures_completes[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "LineAmount",
         "CustomerID", "Country"]
    ]
)
```

```python
# Identifier les 819 factures concernées
factures_prix_nuls = df_prix_nuls_restants["InvoiceNo"].unique()

# Récupérer toutes les lignes de ces factures
df_analyse_globale = df[
    df["InvoiceNo"].isin(factures_prix_nuls)
].copy()

# Calculer les indicateurs par facture
analyse_globale = (
    df_analyse_globale
    .groupby("InvoiceNo")
    .agg(
        lignes_prix_nul=("UnitPrice", lambda x: (x == 0).sum()),
        lignes_prix_positif=("UnitPrice", lambda x: (x > 0).sum()),
        lignes_prix_negatif=("UnitPrice", lambda x: (x < 0).sum())
    )
)

# Classer les factures
factures_mixtes = (
    (analyse_globale["lignes_prix_nul"] > 0)
    & (analyse_globale["lignes_prix_positif"] > 0)
)

factures_zero_uniquement = (
    (analyse_globale["lignes_prix_nul"] > 0)
    & (analyse_globale["lignes_prix_positif"] == 0)
)

print("ANALYSE GLOBALE DES FACTURES")
print("Total des factures :", len(analyse_globale))
print("Factures mixtes :", factures_mixtes.sum())
print("Factures sans ligne à prix positif :", factures_zero_uniquement.sum())
print("Factures avec ligne à prix négatif :",
      (analyse_globale["lignes_prix_negatif"] > 0).sum())
```

```python

# 1. Conserver uniquement les transactions associées à un client identifié
df_clients = df_ventes.loc[
    df_ventes["CustomerID"].notna()
].copy()

# 2. Calculer le CA brut et les annulations par client
ca_clients = (
    df_clients
    .groupby(["CustomerID", "TransactionType"])["LineAmount"]
    .sum()
    .unstack(fill_value=0)
)

# Garantir la présence des deux catégories
ca_clients = ca_clients.reindex(
    columns=["VENTE_STANDARD", "ANNULATION_COMMERCIALE"],
    fill_value=0
)

# 3. Calculer le CA net par client
ca_clients["CA_net"] = (
    ca_clients["VENTE_STANDARD"]
    + ca_clients["ANNULATION_COMMERCIALE"]
)

# 4. Calculer le nombre de factures de vente par client
factures_clients = (
    df_clients.loc[
        df_clients["TransactionType"] == "VENTE_STANDARD"
    ]
    .groupby("CustomerID")
    .agg(
        nombre_factures=("InvoiceNo", "nunique"),
        quantite_vendue=("Quantity", "sum")
    )
)

# 5. Regrouper les indicateurs dans un tableau
analyse_clients = ca_clients.join(factures_clients)

# 6. Trier les clients par CA net décroissant
top_10_clients = (
    analyse_clients
    .sort_values("CA_net", ascending=False)
    .head(10)
    .reset_index()
)

display(top_10_clients)
```

```python

# 1. Regrouper les ventes par pays et type de transaction
ca_pays = (
    df_ventes
    .groupby(["Country", "TransactionType"])["LineAmount"]
    .sum()
    .unstack(fill_value=0)
)

# Garantir la présence des deux catégories
ca_pays = ca_pays.reindex(
    columns=["VENTE_STANDARD", "ANNULATION_COMMERCIALE"],
    fill_value=0
)

# 2. Calculer le CA net par pays
ca_pays["CA_net"] = (
    ca_pays["VENTE_STANDARD"]
    + ca_pays["ANNULATION_COMMERCIALE"]
)

# 3. Calculer les autres indicateurs par pays
indicateurs_pays = (
    df_ventes.loc[
        df_ventes["TransactionType"] == "VENTE_STANDARD"
    ]
    .groupby("Country")
    .agg(
        nombre_factures=("InvoiceNo", "nunique"),
        nombre_clients=("CustomerID", "nunique"),
        quantite_vendue=("Quantity", "sum")
    )
)

# 4. Regrouper les indicateurs
analyse_pays = ca_pays.join(indicateurs_pays)

# 5. Classer les pays par CA net décroissant
top_10_pays = (
    analyse_pays
    .sort_values("CA_net", ascending=False)
    .head(10)
    .reset_index()
)

display(top_10_pays)
```
